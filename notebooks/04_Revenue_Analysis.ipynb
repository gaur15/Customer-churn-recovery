# 💰 Revenue Analysis & Recovery Plan
Hypothesis testing to quantify revenue at risk + prioritised retention offers.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
from src.data_preprocessing import load_data, clean_data
from src.hypothesis_testing import run_all_tests
from src.revenue_recovery import build_recovery_plan
from src.model import load_model

df = clean_data(load_data('../data/processed/customer_data.csv'))
```

```python
results = run_all_tests(df)
for name, r in results.items():
    print(f"{r['test']}  |  p={r['p_value']}  |  Reject H0: {r['reject_h0']}")
```

```python
# Revenue at risk estimate
churned_clv  = df.loc[df['churn']==1, 'clv'].mean()
retained_clv = df.loc[df['churn']==0, 'clv'].mean()
print(f'Avg CLV Churned : ₹{churned_clv:,.2f}')
print(f'Avg CLV Retained: ₹{retained_clv:,.2f}')
print(f'Revenue at risk : ₹{(retained_clv - churned_clv) * (df.churn==1).sum():,.2f}')
```

```python
# Recovery plan (using random scores for demo — in production use trained model)
np.random.seed(42)
n = 1000
recovery_df = build_recovery_plan(
    customer_ids=df['customer_id'].sample(n).values,
    churn_proba=np.random.uniform(0.3, 0.95, n),
    clv=df['clv'].sample(n).values
)
recovery_df.groupby('tier')[['recovery_score','projected_recovery']].sum().round(2)
```
