# 🔧 Feature Engineering
RFM scores, engagement ratios, tenure buckets, and risk flags.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
from src.data_preprocessing import load_data, clean_data
from src.feature_engineering import engineer_features

df = clean_data(load_data('../data/processed/customer_data.csv'))
df = engineer_features(df)
print(f'Features after engineering: {df.shape[1]}')
df[['customer_id','rfm_score','composite_risk_score','support_to_login_ratio','churn']].head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.groupby('rfm_score')['churn'].mean().plot(kind='bar', ax=axes[0], color='#3B82F6')
axes[0].set_title('Churn Rate by RFM Score')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_: f'{y:.0%}'))

df.groupby('composite_risk_score')['churn'].mean().plot(kind='bar', ax=axes[1], color='#EF4444')
axes[1].set_title('Churn Rate by Composite Risk Score')
axes[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_: f'{y:.0%}'))
plt.tight_layout()
```
