# 🌲 Model Training — Random Forest Classifier
Hyperparameter tuning with GridSearchCV, evaluation with SHAP.

```python
import sys; sys.path.insert(0, '..')
from src.data_preprocessing import load_data, clean_data, encode_categoricals, split_and_scale, validate_schema
from src.feature_engineering import engineer_features
from src.model import train_model, evaluate_model, save_model, explain_with_shap

raw = load_data('../data/processed/customer_data.csv')
validate_schema(raw)
df = encode_categoricals(engineer_features(clean_data(raw)))
data = split_and_scale(df)
print('Data ready')
```

```python
model = train_model(data['X_train'], data['y_train'], data['X_val'], data['y_val'], fast=True)
print('Training complete')
```

```python
metrics = evaluate_model(model, data['X_test'], data['y_test'], data['feature_names'])
print(metrics)
```

```python
explain_with_shap(model, data['X_test'], data['feature_names'])
```

```python
save_model(model, '../models/random_forest_churn.pkl')
print('Model saved')
```
