```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='darkgrid')
%matplotlib inline
```

```python
import os, subprocess
if not os.path.exists('../data/processed/customer_data.csv'):
    subprocess.run(['python', '../data/generate_data.py'])
df = pd.read_csv('../data/processed/customer_data.csv')
print(f'Shape: {df.shape}  |  Churn rate: {df.churn.mean():.2%}')
df.head()
```

```python
df.describe().T.style.background_gradient(cmap='Blues')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.groupby('contract_type')['churn'].mean().sort_values(ascending=False).plot(kind='bar', ax=axes[0], color='#EF4444')
axes[0].set_title('Churn Rate by Contract Type')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_: f'{y:.0%}'))
for cv, lbl, c in [(0,'Retained','#22C55E'),(1,'Churned','#EF4444')]:
    df.loc[df['churn']==cv,'tenure_months'].hist(bins=40, alpha=0.6, label=lbl, color=c, density=True, ax=axes[1])
axes[1].set_title('Tenure Distribution')
axes[1].legend()
plt.tight_layout()
```

```python
cols = ['tenure_months','monthly_charges','support_tickets','payment_delays','satisfaction_score','clv','churn']
plt.figure(figsize=(10,7))
sns.heatmap(df[cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
```
